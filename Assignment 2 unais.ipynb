# Q3. Set Operations on Course Enrollments


```python
course_a = {"John", "Alice", "Bob", "David"}
course_b = {"Alice", "Eve", "Charlie", "David"}
```

```python
only_in_a = course_a.difference(course_b)
only_in_one = course_a.symmetric_difference(course_b)
```

```python
("Students in Course A but not in Course B:", only_in_a)
```

```python
("Students only in one of the two courses:", only_in_one)
```

# Q6. Set Comparison


```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}
```

```python
common = A.intersection(B)
("Common elements:", common)
```

```python
only_a = A.difference(B)
("Elements only in Set A:", only_a)
```

```python
only_b = B.difference(A)
("Elements only in Set B:", only_b)
```
